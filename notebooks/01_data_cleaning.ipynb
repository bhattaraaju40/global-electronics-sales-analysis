```python
products['Unit Cost USD'] = (products['Unit Cost USD'].str.replace('$','', regex = False)
.str.replace(',','', regex= False).astype(float))
```

```python
# Making all columns names to lower case removing space and changing space between to underscore 
def clean_column_names(df):
    df.columns = (df.columns.str.strip().str.lower().str.replace(" ","_"))
    return df
```

```python
customers = clean_column_names(customers)
products = clean_column_names(products)
sales = clean_column_names(sales)
stores = clean_column_names(stores)
```

```python
print(customers.columns.tolist())
print(products.columns.tolist())
print(sales.columns.tolist())
print(stores.columns.tolist())
```

```python
# let change the name little more to look nice on some key things
customers = customers.rename(columns={
    "customerkey": "customer_key"
})

products = products.rename(columns={
    "productkey": "product_key",
    "subcategorykey": "subcategory_key",
    "categorykey": "category_key"
})

sales = sales.rename(columns = {
    "customerkey" : "customer_key",
    "storekey" : "store_key",
    "productkey" : "product_key"
})

stores = stores.rename(columns={
    "storekey": "store_key"
})
```

```python
print(customers.columns.tolist())
print(products.columns.tolist())
print(sales.columns.tolist())
print(stores.columns.tolist())
```
